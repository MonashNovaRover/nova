### Basic Open3D Examples

This notebook aims to show how Open3D can be used to render objects and pointclouds.

```python
import numpy as np
import open3d as o3d
```

```python
# create a mesh box for visualizing
mesh_box = o3d.geometry.TriangleMesh.create_box(width=1.0, height=1.0, depth=1.0)
mesh_box.compute_vertex_normals()
mesh_box.paint_uniform_color([0.9, 0.1, 0.1])
```

```python
# create a coordinate frame for viewing and view the objects
mesh_frame = o3d.geometry.TriangleMesh.create_coordinate_frame(size=0.6, origin=[-2, -2, -2])


# this line will open a new window. You must close it to go to the next cell
o3d.visualization.draw_geometries([mesh_box])
```
